```python
import json
import random

def split_dataset(dataset_path="evrp_dataset.json", seed=42):
    random.seed(seed)
    
    with open(dataset_path, 'r') as f:
        data = json.load(f)
    
    # Group by type
    groups = {'C': [], 'R': [], 'RC': []}
    for key, val in data.items():
        iid = val['instance_id'].upper()
        if iid.startswith('RC'): groups['RC'].append(key)
        elif iid.startswith('R'): groups['R'].append(key)
        elif iid.startswith('C'): groups['C'].append(key)
    
    train_keys, test_keys = [], []
    
    for t, keys in groups.items():
        random.shuffle(keys)
        test_keys.extend(keys[-1:])    # 5 for test
        train_keys.extend(keys[:-2])   # 15 for train
    
    train_data = {k: data[k] for k in train_keys}
    test_data = {k: data[k] for k in test_keys}
    
    # Save
    with open('train_dataset.json', 'w') as f:
        json.dump(train_data, f)
    with open('test_dataset.json', 'w') as f:
        json.dump(test_data, f)
    
    print(f"Train: {len(train_data)} instances (C:{sum('C' in data[k]['instance_id'].upper() and not 'RC' in data[k]['instance_id'].upper() for k in train_keys)}, R:{sum(data[k]['instance_id'].upper().startswith('R') and not 'RC' in data[k]['instance_id'].upper() for k in train_keys)}, RC:{sum('RC' in data[k]['instance_id'].upper() for k in train_keys)})")
    print(f"Test: {len(test_data)} instances (5 per type)")
    
    return train_keys, test_keys

# Run
train_keys, test_keys = split_dataset()
```

```python
import json
import random

def process_and_split(dataset_path="evrp_dataset.json", seed=42):
    random.seed(seed)

    # Load dataset
    with open(dataset_path, 'r') as f:
        data = json.load(f)

    # --------------------------------------
    #   STEP 1: REDUCE VEHICLES TO 5 EACH
    # --------------------------------------
    for key, val in data.items():
        vehicles = val.get("vehicles", {})
        # Keep only vehicles 0–4 (if they exist)
        limited = {str(i): vehicles[str(i)] for i in range(50) if str(i) in vehicles}
        val["vehicles"] = limited

    # --------------------------------------
    #   STEP 2: GROUP INSTANCES BY TYPE
    # --------------------------------------
    groups = {'C': [], 'R': [], 'RC': []}

    for key, val in data.items():
        iid = val['instance_id'].upper()
        if iid.startswith("RC"):
            groups['RC'].append(key)
        elif iid.startswith("R"):
            groups['R'].append(key)
        elif iid.startswith("C"):
            groups['C'].append(key)

    train_keys, test_keys = [], []

    # --------------------------------------
    #   STEP 3: SPLIT (last 2 per group → test)
    # --------------------------------------
    for t, keys in groups.items():
        random.shuffle(keys)
        test_keys.extend(keys[-2:])    # last 2 → test
        train_keys.extend(keys[:-5])   # rest → train

    # Create datasets
    train_data = {k: data[k] for k in train_keys}
    test_data = {k: data[k] for k in test_keys}

    # --------------------------------------
    #   STEP 4: SAVE OUTPUT
    # --------------------------------------
    with open('train_dataset.json', 'w') as f:
        json.dump(train_data, f, indent=2)

    with open('test_dataset.json', 'w') as f:
        json.dump(test_data, f, indent=2)

    print(f"Train: {len(train_data)} instances")
    print(f"Test: {len(test_data)} instances (2 per type)")
    print("Each instance now has exactly 5 vehicles.")

    return train_keys, test_keys


# Run the pipeline
train_keys, test_keys = process_and_split()
```
